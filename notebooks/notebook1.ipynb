```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import pymupdf
from src.parse import PDF_DIR, TEXT_DIR, clean_text

for pid in ["1908.03949", "2212.12782", "2504.14791"]:
    doc = pymupdf.open(PDF_DIR / f"{pid}.pdf")
    raw = "\n".join(p.get_text("text", sort=True) for p in doc)
    clean = (TEXT_DIR / f"{pid}.txt").read_text(encoding="utf-8")
    print(f"{pid}: raw {len(raw):,} -> clean {len(clean):,} ({len(clean)/len(raw):.0%}), "
          f"[EQ] count {clean.count('[EQ]')}")
```

```python
import json, random, re
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
chunks = [json.loads(l) for l in (ROOT / "data" / "processed" / "chunks.jsonl").open(encoding="utf-8")]

for c in random.sample(chunks, 3):
    print(c["chunk_id"], "\n", c["text"], "\n")
ref_like = [c for c in chunks if len(re.findall(r"Phys\. Rev|arXiv:|\bet al\.,", c["text"])) >= 3]
print(len(ref_like), "chunks look like bibliography")
```
